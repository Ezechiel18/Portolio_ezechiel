#Analyse spatiale de l'accessibilité à l'eau  dans la commune Ogou 1 au Togo

L’objectif principal de l’étude est d’analyser la répartition et l’accessibilité aux infrastructures 
d’eau potable dans la commune d’Ogou 1 afin d’identifier les inégalités existantes et de 
proposer des priorités d’action pour améliorer l’accès.

###Approche méthodologique

L’étude s’appuie sur une démarche multi-échelle et multicritère pour analyser l’accès à l’eau potable dans la commune d’Ogou 1. Deux approches principales ont été utilisées :
<br>
<b>Analyse statistique par unités administratives </b>

<ol>
<li> Utilisation du score Z pour comparer le niveau d’équipement en points d’eau entre cantons.</li>

<li>Identification des cantons sous-dotés, moyens ou bien desservis.</li>
</ol>
<br>
<b>Analyse spatiale fine par grille régulière</b>

<ol>
<li>Création d’une grille hexagonale de 500 m × 500 m pour dépasser les limites des cantons </li>

<li> Calcul de la densité de points d’eau et de la proportion de bâtiments ayant accès dans un rayon de 500 m (norme humanitaire ONU).</li>

<li>Reclassification des niveaux d’accès (absence, faible, moyen, bon, total).</li>
</ol>
<br>
<b>Analyse multicritère pour identifier les zones prioritaires</b>
<ol>
<li> Sélection de 5 critères (Niveau d’accessibilité à l’échelle des mailles de 500 m 
× 500 m , Taille population par canton 
Part des zones habitées à accès total, Proportion de ménages à moins de 500 m par canton )
</li>
<li> Attribution de scores, normalisation, pondération et agrégation spatiale.</li>

<li> Production d’une carte de hiérarchisation des zones prioritaires pour de futures infrastructures.</li>
</ol>
<i> Cette approche combinée permet de croiser les échelles, d’éviter les biais des limites administratives et de produire une analyse fine, cohérente et opérationnelle de l’accès à l’eau potable. <i>

###Résultats

Les résultats de ce mémoire contribuent à la prise de connaissance de l’accessibilité physique à l’eau potable dans la commune d’Ogou 1. Les resulats phares sont: l'équipement en infrastructure d'eau potable par grille de 500 * 500 mètres, le taux d'accès des menages à l'échelle d'une grille (nombre de menage se situant à 500 mètre d'un point d'eau * 100/ nombre total de ménages à l'echelle de la grille), l'hierarchisation par priotité des zones  aménager (où on doit renforcer l'accès à l'eau).

 <b> L'équipement en infrastructure d'eau potable par grille de 500 * 500 mètres </br>

La carte ci-dessous  propose une lecture fine du niveau d’équipement en points d’eau potable à travers une analyse par densité réalisée sur une trame régulière de mailles de 500 m x 500 m couvrant l’ensemble de la commune d’Ogou 1.

considérés comme ayant accès). 
La formule suivante est utilisé pour calculer la part d'accessiblité des ménges à l'échellle des griles.
<p style="color: blue;">
PAG = (Nombre de bâtiments ayant accès / Nombre total de bâtiments) × 100

Visuellement ces resultats chiffrés ne sont pas parlant pour un non spécialiste; Il faut donc faire un post-traitement par reclassification afin que les resultats puissent être facile à lire sur carte.

<b> Post -traitement </b>

Pour y arriver, le code ci-dessous  a été exécuté dans le calculatruc des champs de Qgis.

```python
CASE 
WHEN  "Prop_acc_2"  IS NULL THEN 'Absence de bâtiment' 
WHEN  "Prop_acc_2"  = 0 THEN '0 % - Pas d\'accès' 
WHEN  "Prop_acc_2"  > 0 AND  "Prop_acc_2"  <= 50 THEN '1 à 50 % - Faible accès' 
WHEN  "Prop_acc_2"  > 50 AND  "Prop_acc_2"  <= 75 THEN '51 à 75 % - Accès moyen' 
WHEN  "Prop_acc_2"  > 75 AND  "Prop_acc_2" < 100 THEN '76 à 99 % - Bon accès' 
WHEN  "Prop_acc_2"  = 100 THEN '100 % - Accès total' 
END
```

<b> On obtient donc </b>


• 0 % (aucun ménage couvert) 
<br>
• 1 à 25 % (très faible accès des ménages) 
<br>
• 26 à 50 % (faible accès) 
<br>
• 51 à 75 % (accessibilité modérée) 
<br>
• 76 à 99 % (bonne accessibilité) 
<br>
• 100 % (tous les bâtiments couverts) 
<br>
• NULL (absence de bâtiment)

#####Génération du SQL

```python
CREATE TABLE Route(
   Id_route VARCHAR(50),
   Nom_route VARCHAR(50),
   Type_route VARCHAR(20),
   Type_bitume VARCHAR(30),
   Etat_route VARCHAR(50),
   Catégorie VARCHAR(50),
   PRIMARY KEY(Id_route)
);

CREATE TABLE Région(
   Id_region VARCHAR(50),
   Nom_region VARCHAR(50),
   PRIMARY KEY(Id_region)
);

CREATE TABLE Préfecture(
   Id_prefecture VARCHAR(50),
   Nom_prefecture VARCHAR(50),
   Id_region VARCHAR(50) NOT NULL,
   PRIMARY KEY(Id_prefecture),
   FOREIGN KEY(Id_region) REFERENCES Région(Id_region)
);

CREATE TABLE Commune(
   Id_commune VARCHAR(50),
   Nom_commune VARCHAR(50),
   Id_prefecture VARCHAR(50) NOT NULL,
   PRIMARY KEY(Id_commune),
   FOREIGN KEY(Id_prefecture) REFERENCES Préfecture(Id_prefecture)
);

CREATE TABLE Canton(
   Id_canton VARCHAR(50),
   Nom_canton VARCHAR(50),
   Surface_km2_canton DECIMAL(15,2),
   Id_commune VARCHAR(50) NOT NULL,
   PRIMARY KEY(Id_canton),
   FOREIGN KEY(Id_commune) REFERENCES Commune(Id_commune)
);

CREATE TABLE Localité(
   Id_Localité VARCHAR(50),
   Nom_localité VARCHAR(30),
   Type_localite VARCHAR(30),
   Id_canton VARCHAR(50) NOT NULL,
   PRIMARY KEY(Id_Localité),
   FOREIGN KEY(Id_canton) REFERENCES Canton(Id_canton)
);

CREATE TABLE Bâtiment(
   Id_batiment VARCHAR(50),
   Superficie_km2_bat DECIMAL(15,2),
   Id_Localité VARCHAR(50) NOT NULL,
   PRIMARY KEY(Id_batiment),
   FOREIGN KEY(Id_Localité) REFERENCES Localité(Id_Localité)
);

CREATE TABLE Point_d_eau(
   Id_pe VARCHAR(50),
   Type_pe VARCHAR(30),
   Statut_pe VARCHAR(30),
   Propriété VARCHAR(50),
   Organe_gestion VARCHAR(30),
   Id_Localité VARCHAR(50) NOT NULL,
   PRIMARY KEY(Id_pe),
   FOREIGN KEY(Id_Localité) REFERENCES Localité(Id_Localité)
);

CREATE TABLE relie(
   Id_pe VARCHAR(50),
   Id_route VARCHAR(50),
   PRIMARY KEY(Id_pe, Id_route),
   FOREIGN KEY(Id_pe) REFERENCES Point_d_eau(Id_pe),
   FOREIGN KEY(Id_route) REFERENCES Route(Id_route)
);
```
