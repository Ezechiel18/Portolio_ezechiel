# Explication des deux scripts

Ce répository aborde les questions d'automatisation de tâches SIG via le langage python. Le choix est porté sur  l'impact d'un projet routier sur la végétation.

Le script [Impact_analysis_tool](Script_python/Impact_analysis_tool.py) est un outil crée pour étudier  les effets d'un projet routier sur la végétation. Il peut être inséré dans le logiciel Arcgis pro (dans la partie toolbox) et être  modifier selon l'usage.

Le script [Impact_analysis_iterator](Script_python/Impact_analysis_iterator.py) comporte des boucles permettant d'itérer sur plusieurs couches en faisant la même analyse que dans le premier cas.
