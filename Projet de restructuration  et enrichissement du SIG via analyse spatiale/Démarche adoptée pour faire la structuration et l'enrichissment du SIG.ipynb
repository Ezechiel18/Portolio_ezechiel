<h1> Restructuration et enrichissement du Système d'Information Géographique de Cérès Flore </h1>
______________________________________________________________________

Dans ce fichier je présenterai brièvement la démarche adoptée pour réorganiser le SIG de Cérès Flore et les outils mis en place pour garantir la performance de ce SIG pour un meilleur fonctionnement dans le futur. Ceci en partant du diagnostic du SIG jusqu' aux analyses spatiales en passant par la proposition des modèles de données, la création du projet Qfield/Qfield Cloud et l'enrichisssement du SIG.

## Diagnostic du SIG

L’entreprise gère un double flux informationnel. D’une part, elle collecte des données qui alimentent et enrichissent son système d'information et (données entrantes), d’autre part, elle assure la diffusion d'informations vers sa clientèle (données sortantes). Dans ce cadre, elle mobilise à la fois des données géographiques et non géographiques, et utilise différents outils pour répondre à ses besoins, notamment un outil interne <b> (dont je me dois de taire le nom) </b> pour la facturation, Excel pour certaines opérations de suivi ou de traitement, et Qgis pour organiser les données spatiales et produire les cartes.

<h4> Données entrantes </h4>

Il s'agit des données d’intérêt commercial, données sur les types d’espaces naturels  (ZNIEFF, ZICO, Site Natura 2000 , Sites RAMSAR) et des données administratives.

<h4> Données sortantes </h4>

Très peu d’information sort du système d’information de l’entreprise. L'attestation d’origine fait parti de ces informations. Elle constitue "la carte d'identité" des espèces commercialisées par l'entreprise. Elle renseigne non seulement sur les conditions bio-climatiques du site de prélèvement des végétaux mais aussi sur les conditions de conservations des grains.
