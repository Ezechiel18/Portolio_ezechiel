# Identification des  zones  optimales pour implanter les infrastrucures éoliennes

L'objectif de cette analyse es d'identifier les zone où il est  possibe d'implanter les infrastrucutures éoliennes. Cette analyse se base donc sur la règlementation française.
<br>
Pour faciliter  la compréhension, ce document succincte est organisé comme suit: Analyse à réaliser, approche méthodologique, documentation mobilisée et les limites.

## 1. Analyse à réaliser

L’objectif est d’identifier les zones optimales pour implanter des infrastructures éoliennes.
Pour cela, l’analyse doit respecter la réglementation du Code de l’environnement, notamment la Section 11 : Éoliennes (Articles L515-44 à L515-46) et l’article L553-1, qui interdisent l’implantation d’éoliennes à moins de 500 m des habitations, ainsi que dans les zones soumises à des contraintes aéronautiques civiles ou militaires, ou proches des routes et voies ferrées.
La couche périmètre habité est donc utilisée pour intégrer cette contrainte.

Afin de garantir une implantation dans des zones présentant un fort potentiel de vent, la couche gisement de vent à 160 m d’altitude est également mobilisée.
Enfin, pour tenir compte de l’existant, la couche Localisation des mâts éoliens est intégrée à l’analyse.

## 2. Approche méthodologique

L’outil exclut d’abord les périmètres habités, puis ne conserve que les zones où la vitesse du vent est supérieure ou égale à 7 m/s à 160 m d’altitude, avec une zone d’influence de 3 km autour des points à fort potentiel.
Les zones situées à proximité des éoliennes existantes sont aussi exclues : une distance de 2 km autour des mâts a été retenue, en cohérence avec les espacements observés (2 km en longueur, 3 km en largeur).

L’analyse combine ainsi l’ensemble de ces critères. Elle est réalisée sur le département de la Gironde, mais l’outil peut être utilisé sur n’importe quel territoire.
