# Cartographie web sur l'accessibilité de l'eau  dans la commune ogou 1 au Togo.

Ce repository décris le process de visualisation sur une carte web  des réslitats emmanant de l'analyse de l'accessiblité à l'eau dans la commune.

## Pourquoi faire une carte web sur cette thématique ?

Le besoins vient du fait que dans le cadre de cette analyse spatiale,  plusieurs types d'analyses ont été effectuées. Cette carte web est réalisée dans  une logique de visualiser toutes ces réslitats dans un même rendu et d'avoir la possibilité de switcher d'une information (couche d'information) à une autre aisément. 

## Les couches mobilisées

L'analyse proprement dite est plus détaillée dans mon repository [Analyse spatiale de l'accessibilité à l'eau](https://github.com/Ezechiel18/Portolio_ezechiel/tree/main/Analyse%20spatiale%20de%20l'accessibilit%C3%A9%20%C3%A0%20l'eau)

Les couches mobilisées pour faire cette visualisation web:
<ul> 
<li> Hiérarchie des zones prioritaires à aménager</li>
<li> Taux d'accessibilité des popliations à l'eau à l'échelle des grilles</li>
<li> Taux d'accessibilité des popliations à l'eau à l'échelle des cantons</li>
<li> Equipement des points d'eau par grille </li>
<li> Equipement des points d'eau par canton </li>
</ul>
